# 01 — Training preparation
The runnable baseline is now implemented. Continue with `03_train_baseline.ipynb` and `../TRAINING.md`. This notebook inspects the configuration and current app catalog; it does not run training.

```python
from pathlib import Path
import json
import random

root = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / 'configs/baseline.json').is_file() and (p / 'src/adaptive_styles').is_dir()), None)
if root is None:
    raise RuntimeError('Start Jupyter from research/adaptive-styles.')
config = json.loads((root / 'configs/baseline.json').read_text(encoding='utf-8'))
random.seed(config['seed'])
print(json.dumps(config, indent=2))
```

```python
catalog_path = (root / config['app_style_catalog']).resolve()
catalog = json.loads(catalog_path.read_text(encoding='utf-8'))['styles']
styles = {int(key): value['name'] for key, value in catalog.items()}
print('Current color Style IDs:', styles)
print('Parameter bounds:', config['parameter_bounds'])
print('Prompt descriptions are a baseline, not supervised labels.')
```

```python
required = [config['prepared_manifest'], config['labels']]
missing = [path for path in required if not (root / path).is_file()]
print('Missing preparation files:', missing or 'none')
print('Use train.py or notebook 03 to generate candidate preferences and train. Review labels before assessing quality.')
```
